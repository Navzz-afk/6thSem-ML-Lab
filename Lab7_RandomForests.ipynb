```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder

# 1. Load the Iris dataset from 'iris.csv'
try:
    iris_df_new = pd.read_csv('/content/iris (2) - iris (2).csv')
except FileNotFoundError:
    print("iris.csv not found in /content/. Attempting to load from current directory.")
    iris_df_new = pd.read_csv('iris (2) - iris (2).csv')
print("Dataset head:")
display(iris_df_new.head())
print("\nDataset Info:")
iris_df_new.info()

X = iris_df_new.drop('species', axis=1)
y_species = iris_df_new['species']

le = LabelEncoder()
y = le.fit_transform(y_species)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 2. Build a Random Forest (RF) classifier with default n_estimators (10)
rf_default = RandomForestClassifier(n_estimators=10, random_state=42)
rf_default.fit(X_train, y_train)

y_pred_default = rf_default.predict(X_test)
score_default = accuracy_score(y_test, y_pred_default)
print(f"\nPrediction score with default n_estimators (10): {score_default:.4f}")

# 3. Fine-tune the model by changing number of trees in your classifier
best_score = 0
best_n_estimators = 0

estimators_range = range(1, 201, 10) # From 1 to 200 with step 10

print("\nFine-tuning Random Forest classifier...")
for n_est in estimators_range:
    rf_tuned = RandomForestClassifier(n_estimators=n_est, random_state=42)
    rf_tuned.fit(X_train, y_train)
    y_pred_tuned = rf_tuned.predict(X_test)
    current_score = accuracy_score(y_test, y_pred_tuned)

    if current_score > best_score:
        best_score = current_score
        best_n_estimators = n_est

    print(f"  n_estimators = {n_est:<4}, Score = {current_score:.4f}")

print(f"\nBest prediction score achieved: {best_score:.4f} with {best_n_estimators} trees.")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder

# 1. Load dataset
try:
    df = pd.read_csv('/content/iris (2) - iris (2).csv')
except FileNotFoundError:
    df = pd.read_csv('iris (2) - iris (2).csv')

print("Dataset Head:")
print(df.head())

# 2. Separate features and target
X = df.drop('species', axis=1)
y_species = df['species']

# Encode target labels
le = LabelEncoder()
y = le.fit_transform(y_species)

# 3. Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 4. Train baseline model (10 trees)
rf_10 = RandomForestClassifier(n_estimators=10, random_state=42)
rf_10.fit(X_train, y_train)

y_pred_10 = rf_10.predict(X_test)
score_10 = accuracy_score(y_test, y_pred_10)

print(f"\nAccuracy with 10 trees: {score_10:.4f}")

# 5. Hyperparameter tuning
best_score = 0
best_n_estimators = 0

print("\nTuning Random Forest...")

for n_est in range(10, 201, 10):
    rf = RandomForestClassifier(
        n_estimators=n_est,
        max_depth=2,
        random_state=42
    )

    rf.fit(X_train, y_train)
    y_pred = rf.predict(X_test)
    score = accuracy_score(y_test, y_pred)

    # Use >= to capture latest best
    if score >= best_score:
        best_score = score
        best_n_estimators = n_est

    print(f"n_estimators = {n_est:<3} | Accuracy = {score:.4f}")

print(f"\nBest Accuracy: {best_score:.4f} with {best_n_estimators} trees")

# 6. Cross-validation (better evaluation)
print("\nCross-validation results:")

rf_final = RandomForestClassifier(
    n_estimators=best_n_estimators,
    max_depth=2,
    random_state=42
)

cv_score = cross_val_score(rf_final, X, y, cv=5).mean()

print(f"Average CV Score: {cv_score:.4f}")
```
